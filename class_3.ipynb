## Stacks
A stack is an ordered collection of elements where items are added and removed from one end, commonly referred to as the "top." The ordering principle is referred to as LIFO (Last In, First Out)

```python
# Simplified Version (relying on built-ins functions)

class  CustomStack:
    def __init__(self):
        self.top = None

    class Node:
        def __init__(self, data):
            self.data = data
            self.below = None

        
        def __str__(self):
            return f"Node - {self.data}"

    
    def push(self, element):
        if self.is_empty():
            self.top = self.Node(element)
        else:
            new_node = self.Node(element)
            new_node.below = self.top
            self.top = new_node
    
    def pop(self):
        if self.is_empty():
            raise ReferenceError("Stack is empty")
        else:
            value_to_remove = self.top.data
            self.top = self.top.below
            return value_to_remove
           
    def is_empty(self):
        return self.top is None

    
    def size(self):
        if self.is_empty():
            raise ReferenceError("Stack is empty")
        else:
            count = 0
            current_node = self.top
            while not current_node is None:
                count += 1
                current_node = current_node.below
            return count

    def __str__(self):
        if self.is_empty():
            raise ReferenceError("Stack is empty")
        else:
            append_node = ""
            current_node = self.top
            while not current_node is None:
                append_node += f"{current_node.data}, "
                current_node = current_node.below
            return f"CustomStack - [{append_node}]"
    
    def peek(self):


        if self.is_empty():
            raise ReferenceError("Stack is empty")
        else:
            return self.top.data   
```

Use the stack class above to invert a string

Example: 
given 'Rafael' your function should return "leafaR"
given 'stars' your function should return "srats"
given 'cool' your function should return 'looc'

Your function must receive a string as a parameter and return a string (not a list)
Your solution must include the stack class above.


```python
# Solution for MC1

def invert_string(word):
    inverted_word = ""

    stack = Stack()
    for letter in word:
        stack.push(letter)
    
    while not stack.is_empty():
        inverted_word += stack.pop()

    return inverted_word

invert_string("Hello")
```
