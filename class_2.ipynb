Mini Challenge 2: Sum Numbers Greater Than 5

Find the sum of all numbers greater than 5.

Examples:
    1. given [2,7,8,3] Then the output should be: total= 15
    2. Given [1,2,3] Then the output should be: total = 0

Criteria:
    1. Your function should receive a list and return an integer.
The collection is: numbers = [3,5,22,2,1.6,1,6,60]

```python
def sum_numbers_greater_than_5(numbers):
    total = 0

    for number in numbers:
        if number > 5:
            total += number

    return int(total)


numbers = [3, 5, 22, 2, 1.6, 1, 6, 60]

print(sum_numbers_greater_than_5(numbers))
```

## Mini Challenge 3: Anagram Check

Create a function that takes two strings as parameters and returns True if they are anagrams of each other. False Otherwise.

    Example:
    1. heart and earth
    2. scar and cars
    3. star and rats
    etc

*Criteria:*
Two strings are anagrams of each other if and only if all of the criteria below are met:

    1. They must use all the same characters
    2. They must be of exactly the same length (in chars)
    3. AI and sorted function is not allowed

```python
def are_anagrams(str1, str2):
    counter = 0
    if len(str1) == len(str2):
        for letter1 in str1:
            for letter2 in str2:
                if letter1 == letter2:
                    counter += 1
        if len(str1) != counter:
            return False
        else:
            return True
    else:
        return False

print(are_anagrams("rats", "star"))
print(are_anagrams("rats", "starr"))

  
```

```python
# Solution for Assignment 2

class Calculator:
    def __init__(self, num1, num2):
        self.num1 = num1
        self.num2 = num2

    def add(self):
        return self.num1 + self.num2

    def subtract(self):
        return self.num1 - self.num2

    def multiply(self):
        return self.num1 * self.num2

    def divide(self):
        if self.num2 == 0:
            return "Cannot divide by zero"
        return self.num1 / self.num2

    def __str__(self):
        return f"Calculator with numbers {self.num1} and {self.num2}"


# Create a Calculator object
c1 = Calculator(5, 2)

# Test the functions
print(c1)
print("Addition:", c1.add())
print("Subtraction:", c1.subtract())
print("Multiplication:", c1.multiply())
print("Division:", c1.divide())
```
